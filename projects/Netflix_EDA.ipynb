# Netflix Content Analysis (Business Perspective)

This project analyzes Netflix content data from a business analyst perspective to understand content distribution, growth trends, and regional and genre insights.

## Dataset Overview

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('netflix_titles.csv')
df.head()
```

## Data Cleaning

```
df = df.drop_duplicates()
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['year_added'] = df['date_added'].dt.year
df.info()
```

## Movies vs TV Shows Distribution

```
df['type'].value_counts().plot(kind='bar')
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Content Type')
plt.ylabel('Count')
plt.show()
```

## Content Growth Over Years

```
df['year_added'].value_counts().sort_index().plot(kind='line')
plt.title('Netflix Content Growth Over Years')
plt.xlabel('Year')
plt.ylabel('Titles Added')
plt.show()
```

## Ratings Distribution

```
df['rating'].value_counts().head(10).plot(kind='bar')
plt.title('Top 10 Content Ratings')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.show()
```

## Genre Analysis – Top Content Categories

```
genre_series = df['listed_in'].str.split(', ')
all_genres = genre_series.explode()

top_genres = all_genres.value_counts().head(10)

plt.figure(figsize=(8,4))
top_genres.plot(kind='bar')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Insight:
Drama and International Movies dominate Netflix’s catalog, indicating strong focus on storytelling and regional content.

## Country-wise Content Production

```
country_series = df['country'].dropna().str.split(', ')
all_countries = country_series.explode()

top_countries = all_countries.value_counts().head(10)

plt.figure(figsize=(8,4))
top_countries.plot(kind='bar')
plt.title('Top 10 Content Producing Countries on Netflix')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Insight:
The United States leads content production followed by India and the UK, showing Netflix’s strong presence in high consumption markets.

## Monthly Content Addition Trend

```
df['month_added'] = df['date_added'].dt.month

monthly_trend = df['month_added'].value_counts().sort_index()

plt.figure(figsize=(8,4))
monthly_trend.plot(kind='line', marker='o')
plt.title('Monthly Content Addition Trend')
plt.xlabel('Month')
plt.ylabel('Titles Added')
plt.xticks(range(1,13))
plt.tight_layout()
plt.show()
```

### Insight:
Netflix tends to add more content during certain months which may align with seasonal user demand and marketing campaigns.

## Final Business Summary

- Netflix has more Movies than TV Shows, indicating a strong movie-focused content strategy.
- Rapid content growth after 2016 reflects aggressive platform expansion.
- Drama and International content dominate the catalog.
- The US and India are major contributors to Netflix content supply.
- Monthly release patterns suggest strategic content scheduling.

Overall, the dataset highlights Netflix’s focus on scale, regional diversity, and content variety to support global growth.
